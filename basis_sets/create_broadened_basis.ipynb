```python
from fsl_mrs.utils import mrs_io
from fsl_mrs.utils.preproc.filtering import apodize
import numpy as np
from copy import deepcopy
```

```python

basis = mrs_io.read_basis('basis_sets/basis_3T')

broadened = [
    apodize(x, basis.original_dwell, 0.2) for x in basis.original_basis_array.T
]

broadened = np.stack(broadened).T

print(broadened.shape)
print(basis.original_basis_array.shape)

basis_broadened = deepcopy(basis)
basis_broadened._raw_fids = broadened
basis_broadened.save('basis_sets/basis_broadened_3T')
```
